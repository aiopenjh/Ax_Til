### 카페 메뉴 딕셔너리에서 키(Key)와 값(Value)을 각각 분리하여 추출하는 방법을 실습합니다. 이와 함께 연산 기호를 사용하여 두 데이터 그룹의 교집합, 합집합, 차집합을 직관적으로 계산하는 과정을 실습합니다.

1. "coffee"는 4000, "tea"는 4500의 쌍을 가지는 menu 딕셔너리를 선언하세요.
2. keys 메서드를 사용하여 딕셔너리 내 모든 고유 키 목록을 추출하여 출력하고, values 메서드를 사용하여 모든 값 데이터 그룹을 추출하여 출력하세요.
3. 요소로 1, 2, 3을 가지는 set_a 집합과 3, 4, 5를 가지는 set_b 집합을 각각 정의하세요.
4. 교집합 연산 기호(&)를 사용하여 두 집합의 공통 요소를 도출하고 출력하세요.
5. 합집합 연산 기호(|)와 차집합 연산 기호()를 각각 사용하여 두 그룹의 전체 요소를 합친 결과 및 set_a 그룹에만 존재하는 요소만 남긴 결과를 차례로 출력하세요.

```python
menu = {"coffee": 4000, "tea": 4500}
print(menu.keys())    # 키 목록
print(menu.values())  # 값 목록

set_a = {1, 2, 3}
set_b = {3, 4, 5}

# [팁] 변수를 일일이 새로 정의하지 않고 print() 안에 바로 연산식을 넣으면 훨씬 간결하고 튜플 꼬임도 방지!
print(set_a & set_b)  # 교집합 (&)
print(set_a | set_b)  # 합집합 (|)
print(set_a - set_b)  # 차집합 (-)
```
