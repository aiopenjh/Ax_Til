## 2️⃣ [2장 2강] 기본 자료형과 형변환 (Types & Type Casting)

#### 📝 실습 1. 문자열과 숫자의 결합을 위한 명시적 형변환 (`str()`)
- 정수형 `age`를 `str(age)`로 변환하여 문자열과 에러 없이 결합

```python
name = "최지훈"
age = 37
message = "나는 " + name + " 이고 내 나이는 " + str(age) + "살입니다."
print(message)
```

#### 📝 실습 2. 정수와 실수의 묵시적/명시적 형변환
- `10 + 3.14` 자동 실수 변환(`float`) 및 `int()` 변환 시 소수점 버림 처리 확인

```python
x = 10
y = 3.14
result = x + y
print(result)       # 묵시적 형변환 (float)
print(int(result))  # 명시적 형변환 (소수점 이하 버림)
```

#### 📝 실습 3. 외부 입력 데이터 검증 및 방어적 형변환 (`.isdigit()`)
> **핵심 팁**: `.isdigit()`은 불리언(`True/False`)을 반환하는 메서드이므로 `if user_input.isdigit():` 로 사용

```python
user_input = "25"

# [헷갈렸던 점 1] .isdigit()은 메서드라 () 필수! True/False를 반환하므로 ">= 0" 비교 불필요
# [헷갈렸던 점 2] 파이썬 조건문 끝에는 세미콜론(;)이 아니라 콜론(:)
if user_input.isdigit():
    age = int(user_input)  # 문자열 -> 정수로 안전하게 변환
    print("정수 전환 성공")
else:
    print("숫자만 입력해주세요!")
```

## 3️⃣ [2장 3강] 연산자 (Operators)

#### 📝 실습 1. 대입 연산과 복합 대입
- `count = 10`에 5를 곱해 다시 저장

```python
count = 10
count = count * 5  # 또는 복합 대입: count *= 5
print(count)
```

#### 📝 실습 2. 문자열 크기 비교(`<`)와 아스키/유니코드 원리
> **원리**: 문자열 크기 비교는 글자 수가 아니라 **사전 순서(아스키/유니코드 `ord()`)**로 맨 첫 글자부터 비교!

```python
# [핵심] 문자열 크기 비교는 글자 수가 아니라 '사전 순서(ord() 문자 번호)' 기준!
# 'a'(97) < 'b'(98) 이므로 True
fruits = 'apple' < 'banana'
print(fruits)

# id()는 컴퓨터 RAM 메모리 주소 (크기 비교 기준 아님)
# 파이썬 최적화(인턴/캐싱)로 자주 쓰이는 문자열/작은 숫자는 메모리에 미리 고정되기도 함
print(f"id 메모리 주소: {id('apple')=}, {id('banana')=}")
print(f"ord 고유 번호: ord('a')={ord('a')}, ord('b')={ord('b')}")
```

#### 📝 실습 3. `and` 단락 평가 (Short-Circuit Evaluation)를 통한 에러 방어
> **단락 평가**: `and` 연산 시 좌측이 `False`이면 우측의 0으로 나누기(`ZeroDivisionError`) 식을 아예 실행하지 않음!

```python
# [단락 평가] and는 앞이 False면 뒤의 에러 식(500 % 0)을 아예 실행하지 않고 바로 False 반환!
# (주의: or 연산자는 앞이 False면 뒤까지 계산하므로 ZeroDivisionError 발생)
result = False and 500 % 0

print(result)  # 에러 없이 안전하게 False 출력
```

## 4️⃣ [2장 4강] 문자열 형식화 (String Formatting)

#### 📝 실습 1. 퍼센트(`%`) 연산자 vs 최신 `f-string` 비교
- `%` 방식의 엄격한 타입/순서 규칙 vs `f-string`의 직관성 비교

```python
menu_name = 'Americano'
menu_price = 4500

# [% 방식 (1세대)] %s는 문자열, %d는 정수. 순서 바뀌거나 타입 안 맞으면 TypeError 발생!
print("상품: %s | 가격: %d원" % (menu_name, menu_price))

# [f-string (현대 표준)] 타입 기호 안 따지고 {변수명} 그대로 꽂아 넣어 훨씬 안전하고 직관적
print(f"상품: {menu_name} | 가격: {menu_price}원")
```

#### 📝 실습 2. 너비 확보, 우측 정렬 및 천 단위 쉼표 (`:>8,`)
> **서식 규칙**: `{변수명 : [정렬방향][너비칸수][천단위쉼표]}`

```python
price = 15000
# [f-string 서식 규칙] 콜론(:) 뒤에 [정렬방향][칸수][천단위쉼표] 순으로 작성
# >8,  ==> 오른쪽 정렬(>), 총 8칸 확보, 3자리마다 쉼표(,)
print(f"금액: {price:>8,}")
```

#### 📝 실습 3. 디버깅 전용 등호 조절자 (`{변수명=}`)
- 변수명을 앞뒤로 두 번 쓰지 않고 `{변수명=}` 하나로 변수명과 현재 값을 동시 출력

```python
current_status = '정산 진행중'
price = 6500

# [디버깅 치트키] {변수명=} 형태로 쓰면 변수 이름과 값을 알아서 같이 출력 (앞에 라벨 안 쳐도 됨)
print(f"디버깅 로그 -> {current_status=}, {price=}")
```

## 5️⃣ [3장 1강] 시퀀스 자료구조 (List & Tuple)

#### 📝 실습 1. 리스트 수정(가변성)과 튜플 음수 인덱싱
- 리스트 요소 변경(`fruits[1] = 'blueberry'`) 및 튜플 마지막 요소 추출(`colors[-1]`)

```python
fruits = ['apple', 'banana', 'cherry']
fruits[1] = 'blueberry'
print(fruits)

colors = ('red', 'green', 'blue')
print(colors[-1])
```

#### 📝 실습 2. 리스트 결합(`+`) 및 포함 여부 확인(`in`)
- 두 리스트 병합 및 `in` 연산자를 활용한 권한 검증

```python
group1 = [1, 2]
group2 = [3, 4]
group = group1 + group2
print(group)

user_roles = ['admin', 'editor', 'member']
adminhave = user_roles is not None and 'admin' in user_roles
print(f"{adminhave=}")
```

#### 📝 실습 3. 슬라이싱 및 가변 언패킹 (`*others`)
> **가변 언패킹 규칙**: `*` 별표는 한 줄에 딱 1개만 사용 가능! 첫 번째를 제외한 나머지를 리스트로 묶어줌

```python
numbers = [0, 1, 2, 3, 4, 5]
print(numbers[1:4])  # 슬라이싱 [이상:미만] -> 인덱스 1, 2, 3 추출

numbers2 = [10, 20, 30, 40]
# [가변 언패킹] *별표는 첫 번째를 뺀 나머지를 리스트로 묶어줌
# 팁: *별표는 한 줄에 딱 1개만 가능! 두 번째를 빼려면 (first, second, *others) 또는 (_, second, *others)
first, *others = numbers2
print(first)
print(others)  # *others 대신 변수 그대로 출력하면 [20, 30, 40] 리스트로 출력
```

## 6️⃣ [3장 2강] 매핑 및 집합 자료구조 (Dict & Set)

#### 📝 실습 1. 딕셔너리 동적 수정/추가 및 안전한 조회 (`.get()`)
> **대참사 방지 팁**: 딕셔너리 키는 무조건 따옴표(`"age"`)를 붙여야 함! 따옴표를 빼면 이전 셀의 변수 `age(25)`가 키로 들어감

```python
user_info = {"name": "Sparta", "age": 20}

# [주의!] 따옴표 필수: user_info["age"] = 21
# 따옴표 없이 user_info[age] 쓰면 이전 셀의 변수 age(25)가 키로 들어가서 {25: 21} 참사 발생!
user_info["age"] = 21
user_info["role"] = "admin"

# [.get(키, 기본값)] 대괄호 조회와 달리 키가 없어도 KeyError 에러 없이 안전하게 기본값 반환
print(user_info.get("phone", "010-0000-0000"))
print(user_info)
```

#### 📝 실습 2. 실무 데이터 정제 파이프라인 (`list(set())`)
- `list` ➡️ `set`(중복 제거) ➡️ `list`(순서 및 인덱싱 복원)

```python
raw_emails = ["a@sparta.com", "b@sparta.com", "a@sparta.com"]

# 1단계: 집합(set)으로 변환하여 중복 자동 제거
emails = set(raw_emails)

# 2단계: 슬라이싱과 인덱스 접근을 위해 다시 리스트로 복원
refined_emails = list(emails)

print(refined_emails)
```

#### 📝 실습 3. 딕셔너리 키/값 분리 및 집합 연산자 (`&`, `|`, `-`)
- 불필요한 중간 변수 없이 `print()` 내에서 직관적으로 연산 수행

```python
menu = {"coffee": 4000, "tea": 4500}
print(menu.keys())    # 키 목록
print(menu.values())  # 값 목록

set_a = {1, 2, 3}
set_b = {3, 4, 5}

# [팁] 변수를 일일이 새로 정의하지 않고 print() 안에 바로 연산식을 넣으면 훨씬 간결하고 튜플 꼬임도 방지!
print(set_a & set_b)  # 교집합 (&)
print(set_a | set_b)  # 합집합 (|)
print(set_a - set_b)  # 차집합 (-)
```
